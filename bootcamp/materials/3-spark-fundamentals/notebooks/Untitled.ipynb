```python
events = [
    {"event_id": "e1", "timestamp": "2025-06-24T10:00:00", "user_id": "u1"},
    {"event_id": "e2", "timestamp": "2025-06-24T10:01:00", "user_id": "u2"},
    {"event_id": "e1", "timestamp": "2025-06-24T10:02:00", "user_id": "u3"},
]
```

```python
my_dict={}

def dedup(events):
    for e in events:
            my_dict[e['event_id']] = e

    return(list(my_dict.values()))
```

```python
dedup(events)
```

```python
my_dedup = {}

def dedup_unsorted(events):
    for e in events:
        if e['event_id'] not in my_dedup.keys():    
            my_dedup[e['event_id']] = e
        else:
            ts = my_dedup[e['event_id']]['timestamp']
            if ts < e['timestamp'] :
                my_dedup[e['event_id']] = e
                
    return(list( my_dedup.values()))
```

```python
dedup_unsorted(events)
```
